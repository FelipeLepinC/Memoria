```python
import pandas as pd
import seaborn as sns; sns.set()
import matplotlib.pyplot as plt
```

```python
# drive.mount('/content/drive')
# path = '/content/drive/MyDrive/Memoria_Group_Classification/Datos/'
df = pd.read_csv('APP_CORP_SPLIT_DATA.csv')
print(df.shape)
print(df.head(5))
```

```python
print("\n")
fig, ax = plt.subplots(figsize=(6,3),nrows=1)
sns.countplot(x=df['label'],palette="Blues_d")
ax.set_title('Distribución del los datos',fontsize=9)
ax.tick_params(axis='both', which='major', labelsize=9)
ax.set_xlabel("clases",fontsize=9)
ax.set_ylabel("numero de ejemplos",fontsize=9)
plt.show()
```
